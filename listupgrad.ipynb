a batsman has scored 10, 60, 100, 40, 120 in his career find out no. of matches average run scored minimum run maximum run scored by in the career

```python
runs = [10, 60, 100, 40, 120]

# Calculating career statistics
matches = len(runs)
avg_runs = sum(runs) / matches
min_runs = min(runs)
max_runs = max(runs)

# Displaying stats in a formatted scorecard style
print("=" * 35)
print("      BATSMAN CAREER STATS      ")
print("=" * 35)
print(f"{'-' * 35}")
print(f" Matches Played : {matches:<15}")
print(f" Total Runs     : {sum(runs):<15}")
print(f" Average Runs   : {avg_runs:<15.2f}")
print(f" Minimum Runs   : {min_runs:<15}")
print(f" Maximum Runs   : {max_runs:<15}")
print(f"{'-' * 35}")
print("=" * 35)
```

### 1. Find the longest word in a list of words

We will define our list of words, iterate through it to keep track of the longest word found, and then print it.

```python
words = ['php', 'elephant', 'phython', 'tiger', 'lion']

longest_word = ""
for word in words:
    if len(word) > len(longest_word):
        longest_word = word

print(f"The longest word is: {longest_word}")
```

### 2. Sort a list without using any functions

We will use a standard sorting algorithm, such as **Bubble Sort**, to sort the list `[10, 1, 9, 199]` in ascending order without relying on Python's built-in `sort()` or `sorted()` functions.

```python
a = [10, 1, 9, 199]
n = len(a)

# Bubble Sort algorithm
for i in range(n):
    for j in range(0, n - i - 1):
        if a[j] > a[j + 1]:
            # Swap elements
            a[j], a[j + 1] = a[j + 1], a[j]

print(f"Sorted list: {a}")
```

### 3. Print diagonal elements in a list of lists

For a matrix represented as a list of lists, the diagonal elements are those where the row index equals the column index (i.e., `a[i][i]`).

```python
a = [[100, 200, 300], [10, 20, 30], [1000, 2000, 3000]]

diagonal_elements = []
for i in range(len(a)):
    diagonal_elements.append(a[i][i])

print(f"Diagonal elements: {diagonal_elements}")
```

### 4. Square each element of a list

We can square each element of the list `[20, 21, 30, 40, 45]` efficiently using a list comprehension.

```python
a = [20, 21, 30, 40, 45]
squared_elements = [num ** 2 for num in a]

print(f"Original list: {a}")
print(f"Squared list:  {squared_elements}")
```

### 5. Print unique elements in a list

We can extract unique elements from the list `[20, 1, 20, 30, 40, 1, 2]` while preserving the original order of appearance by keeping track of the seen elements in a temporary list or set.

```python
a = [20, 1, 20, 30, 40, 1, 2]
unique_elements = []

for num in a:
    if num not in unique_elements:
        unique_elements.append(num)

print(f"Original list: {a}")
print(f"Unique list:   {unique_elements}")
```

### 6. Find the mode of a list

To find the mode of a list, we can keep track of the count of each element using a dictionary and then identify the element with the maximum count.

```python
a = [10, 11, 11, 10, 12, 13, 10]

# Dictionary to store frequencies
frequency = {}
for num in a:
    frequency[num] = frequency.get(num, 0) + 1

# Find the element with the highest frequency
mode = None
max_count = 0
for num, count in frequency.items():
    if count > max_count:
        max_count = count
        mode = num

print(f"Original list: {a}")
print(f"Mode: {mode} (appears {max_count} times)")
```

### 7. Remove empty string from list of strings

We can remove empty or whitespace-only strings from the list by using a list comprehension to filter out strings that are empty or contain only spaces.

```python
a = ['virat', 'kohli', 'kl', ' ', 'ishan']

# Filter out empty or whitespace-only strings
cleaned_list = [string for string in a if string.strip() != '']

print(f"Original list: {a}")
print(f"Cleaned list:  {cleaned_list}")
```
