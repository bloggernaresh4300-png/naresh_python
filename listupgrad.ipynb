a batsman has scored 10, 60, 100, 40, 120 in his career find out no. of matches average run scored minimum run maximum run scored by in the career

```python
runs = [10, 60, 100, 40, 120]

# Calculating career statistics
matches = len(runs)
avg_runs = sum(runs) / matches
min_runs = min(runs)
max_runs = max(runs)

# Displaying stats in a formatted scorecard style
print("=" * 35)
print("      BATSMAN CAREER STATS      ")
print("=" * 35)
print(f"{'-' * 35}")
print(f" Matches Played : {matches:<15}")
print(f" Total Runs     : {sum(runs):<15}")
print(f" Average Runs   : {avg_runs:<15.2f}")
print(f" Minimum Runs   : {min_runs:<15}")
print(f" Maximum Runs   : {max_runs:<15}")
print(f"{'-' * 35}")
print("=" * 35)
```

```python
"""
in a list [10,21,23,5,15,22,25] count the even and odd numbers and print the result
"""
```

```python
# Restore the built-in print function in case it was overwritten
try:
    del print
except NameError:
    pass

a = [10, 21, 23, 5, 15, 22, 25]
even_no_count = 0
odd_no_count = 0

for num in a:
    if num % 2 == 0:
        even_no_count += 1
    else:
        odd_no_count += 1

print("even numbers: " + str(even_no_count))
print("odd numbers: " + str(odd_no_count))
```
