```python
import sympy as sp

# Define the variables
x, y = sp.symbols('x y')

# Define the system of equations
eq1 = sp.Eq(x + y, 15)
eq2 = sp.Eq(x - y, 5)

# Solve the equations
solution = sp.solve((eq1, eq2), (x, y))
print(f"The solution is: x = {solution[x]}, y = {solution[y]}")
```
