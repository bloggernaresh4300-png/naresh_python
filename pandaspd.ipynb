# Tata Car Price Prediction - Data Analysis & Aggregation
Is notebook mein hum Tata car price prediction dataset par data exploration, transformations, aggregations (groupby), aur sorting perform kar rahe hain.

```python
import pandas as pd
```

### Step 1: Libraries Import
Sabse pehle hum data manipulation ke liye Python ki standard library `pandas` ko import kar rahe hain.

```python
data = pd.read_csv('/content/tata_car_price_prediction_dataset (1).csv')
```

### Step 2: Dataset Load Karna
Yahan hum Tata car price prediction ki CSV file ko read karke ek Pandas DataFrame `data` mein load kar rahe hain.

```python
data.head()
```

### Step 3: Feature Engineering (Value Ratio)
Hum car ki value density nikalne ke liye ek naya column `'value'` bana rahe hain, jo `ex_showroom_price_lakh` ko `resale_price_lakh` se divide karke ratio calculate karta hai.

```python
data['value'] = data['ex_showroom_price_lakh']/data['resale_price_lakh']
```

### Step 4: Groupby on Fuel Type
Fuel Type ke aadhar par resale price ka mean, ex-showroom price ka sum, aur unique body types ka count nikalne ke liye groupby perform kiya gaya hai.

```python
# Groupby perform karein
grouped_data = data.groupby('fuel_type').agg({
    'resale_price_lakh': 'mean',
    'ex_showroom_price_lakh': 'sum',
    'body_type': 'nunique'
}).reset_index()

# S.No. index set karein (1, 2, 3...)
grouped_data.index = grouped_data.index + 1
grouped_data.index.name = 'S.No.'

# Result display karein
grouped_data
```

```python
# Aggregation logic defined as per user requirement
grouped_df = data.groupby('fuel_type').agg({
    'resale_price_lakh': ['sum', 'max', 'min'],
    'ex_showroom_price_lakh': 'sum',
    'body_type': 'unique'
}).reset_index()

# Multi-index columns ko flatten aur clean karna
grouped_df.columns = [
    '_'.join(col).strip() if isinstance(col, tuple) else col
    for col in grouped_df.columns
]

# S.No. (1, 2, 3...) set karna
grouped_df.index = grouped_df.index + 1
grouped_df.index.name = 'S.No.'

# Output display karna
grouped_df
```

find the average of engine_cc and km_driven for all the combination of transmission and body type


```python
# 1. Groupby transmission and body_type
# 2. Find the average (mean) for engine_cc and kilometers_driven

average_data = data.groupby(['transmission', 'body_type'])[['engine_cc', 'kilometers_driven']].mean().reset_index()

# S.No. set karna (1, 2, 3...)
average_data.index = average_data.index + 1
average_data.index.name = 'S.No.'

# Result ko show karna
average_data
```

### Step 6: Transmission aur Body Type Combination Analysis
Yahan hum combination of `transmission` aur `body_type` ke liye engine capacity (`engine_cc`) aur total kilometers driven (`kilometers_driven`) ka average nikaal rahe hain, jise humne `average_data` variable mein save kiya hai.

```python
data.groupby(['transmission', 'body_type'])[['engine_cc', 'kilometers_driven']].mean()
```

```python
data.groupby(['transmission', 'body_type']).agg({'engine_cc':'mean','kilometers_driven':'mean'}).reset_index()
```

```python
# 'average_data' DataFrame ke columns ko rename karein
a = a.rename(columns = {
    'engine_cc': 'engine_cc_mean',
    'kilometers_driven': 'kilometers_driven_mean'
})

# Result ko show karein
average_data
```
