### **Understanding Pandas `merge()`**

#### **1. Table Caption for Your Merged Results**
* **Caption:** *"Comparison of Transmission-Level Total Kilometers Driven with Body-Type Breakdowns using Left and Right Joins"*

---

#### **2. What is Merging?**
Merging in Pandas is the process of combining two or more DataFrames based on common columns (keys) or indexes. It is highly analogous to **SQL Joins**.

---

#### **3. Why We Use Merging?**
In relational databases and tidy data structures, information is often split across multiple tables to avoid redundancy (normalization). We use merging to bring this distributed information back together into a single DataFrame for analysis, plotting, or modeling.

---

#### **4. Benefits of Merging**
* **Data Alignment:** Automatically aligns rows based on matching keys, preventing mismatch errors.
* **Flexibility:** Supports multiple join types (`inner`, `left`, `right`, `outer`) to control how unmatched keys are handled.
* **Memory Efficiency:** Allows you to store relational tables separately and combine them only when necessary.

#### **5. Simple and Clear Code Example**

Here is a simple example demonstrating a Left Merge between a parent table (Group) and a details table (Sub-Group):

```python
import pandas as pd

# Left table: Total sum per transmission
left_df = pd.DataFrame({
    'transmission': ['Automatic', 'Manual'],
    'total_km': [68986890, 116641178]
})

# Right table: Breakdown by body type
right_df = pd.DataFrame({
    'transmission': ['Automatic', 'Automatic', 'Manual'],
    'body_type': ['SUV', 'Hatchback', 'SUV'],
    'body_km': [9527784, 20274185, 9086719]
})

# Simple left merge on 'transmission'
merged_example = pd.merge(left_df, right_df, on='transmission', how='left')

print("Left DataFrame (Summary):")
display(left_df)
print("\nRight DataFrame (Details):")
display(right_df)
print("\nMerged Result:")
display(merged_example)
```

### **Reverting a Transposed DataFrame**

Transposing (`.T`) swaps rows and columns. Applying `.T` again reverts the DataFrame back to its original shape.

```python
# 1. Original DataFrame
print("--- Original df2 ---")
display(df2.head(3))

# 2. Transposed DataFrame
df2_transposed = df2.T
print("\n--- Transposed (df2.T) ---")
display(df2_transposed.iloc[:, :3])

# 3. Reverting back to original shape by transposing again
df2_reverted = df2_transposed.T
print("\n--- Reverted back to original (df2_transposed.T) ---")
display(df2_reverted.head(3))
```
