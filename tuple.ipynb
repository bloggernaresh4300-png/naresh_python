write a udf to find whether a number is positive or negative

```python
def check_tuple_elements(input_tuple):
    results = []
    for num in input_tuple:
        if num > 0:
            results.append((num, 'Positive'))
        elif num < 0:
            results.append((num, 'Negative'))
        else:
            results.append((num, 'Zero'))
    return results

# Test the function with a tuple
test_tuple = (-5, 10, 0, -3, 8)
print(check_tuple_elements(test_tuple))
```

```python
test_tuple = (-451, 100, 0, -38, 909)
print(check_tuple_elements(test_tuple))
```

### Summary of the Tuple Integer Checker Function

The `check_tuple_elements` function is a User-Defined Function (UDF) designed to analyze a collection of numbers (passed as a tuple):

1. **Input:** A tuple containing positive numbers, negative numbers, or zeros.
2. **Logic:** It loops through each item in the tuple and evaluates whether it is greater than zero, less than zero, or equal to zero.
3. **Output:** It returns a list of tuples containing each original number paired with its corresponding category ('Positive', 'Negative', or 'Zero').

```python
def positive_negative_checker(a):
  if a > 0:
    return 'positive'
  elif a < 0:
    return 'negative'
  else:
    return 'zero'
```

```python
#write a udf to count vowel and consonants in any name ?
```

```python
def count_vowels_and_consonants(name):
    vowels = 'aeiouAEIOU'
    num_vowels = 0
    num_consonants = 0

    for char in name:
        if char.isalpha():
            if char in vowels:
                num_vowels += 1
            else:
                num_consonants += 1

    return {'vowels': num_vowels, 'consonants': num_consonants}

# Test the function
name_to_test = "Virat Kohli"
result = count_vowels_and_consonants(name_to_test)
print(f"Name: {name_to_test}")
print(f"Vowels: {result['vowels']}")
print(f"Consonants: {result['consonants']}")
```

```python
def c_c_counter(name):
  a = {'vowel':0, 'consonanta':0}
  for i in range(len(name)):
    if name[i] in 'aeiouAEIOU':
      a['vowel'] = a['vowel']+1
    else:
      a['consonant'] = a['consonant']+1
    return a
```
