```python
name = 'deepanjan'
freq = {}
for char in name:
    if char in freq:
        freq[char] += 1
    else:
        freq[char] = 1
print(freq)
```

```python
 wap to count no. of vowels and consonants in list of names eg. ['ashish','kirish'] ->
#{'ashish'}:{'vowel':2,'consonants':4},'krish:{'vowel':1,'consonants':4}}
```

```python
names_list = ['ashish', 'kirish']
results = {}

for name in names_list:
    counts = {'vowel': 0, 'consonants': 0}
    for char in name.lower():
        if char.isalpha():
            if char in 'aeiou':
                counts['vowel'] += 1
            else:
                counts['consonants'] += 1
    results[name] = counts

print(results)
```

```python
wap to count frequency of each alphabet in a list of names eg ['ashish','krish'] ->
#{'ashish':'a':1,'s':2,'h':1,'i':1}
```

```python
names = ['ashish', 'krish']
result_freq = {}

for name in names:
    char_freq = {}
    for char in name:
        if char.isalpha():
            if char in char_freq:
                char_freq[char] += 1
            else:
                char_freq[char] = 1
    result_freq[name] = char_freq

print(result_freq)
```
