```python
# Code to print a Kajukatli (Diamond) pattern of stars
n = 6  # This represents the size/half-height of the diamond

# Upper part of the diamond
for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)

# Lower part of the diamond
for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " * i)
```

```python
# Code to draw Doraemon's face using only stars (*)

# Defining the matrix dimensions
height = 25
width = 50

for y in range(height):
    row_chars = []
    for x in range(width):
        # Normalize coordinates so the center of the face is around (25, 12)
        nx = (x - 25) / 1.7
        ny = (12 - y)

        # 1. Main Head Outline (Radius approx 10)
        head_val = nx**2 + ny**2
        if 9.5 <= head_val <= 11.5:
            row_chars.append("*")
            continue

        # 2. Eyes (Two small ellipses near the top center)
        left_eye = ((nx + 2)**2)/1.2 + ((ny - 4.5)**2)/1.8
        right_eye = ((nx - 2)**2)/1.2 + ((ny - 4.5)**2)/1.8
        if (0.8 <= left_eye <= 1.2) or (0.8 <= right_eye <= 1.2):
            row_chars.append("*")
            continue

        # Pupils (Inner eye dots)
        if (left_eye < 0.3) or (right_eye < 0.3):
            row_chars.append("*")
            continue

        # 3. Nose (Small circle centered right below the eyes)
        nose_val = nx**2 + (ny - 2)**2
        if 0.5 <= nose_val <= 1.2:
            row_chars.append("*")
            continue

        # 4. Vertical line from nose to mouth
        if abs(nx) < 0.3 and -4.5 <= ny <= 1.1:
            row_chars.append("*")
            continue

        # 5. Smiling Mouth (Arc at the bottom)
        mouth_val = nx**2 + (ny + 1)**2
        if 15.0 <= mouth_val <= 17.0 and ny < -1:
            row_chars.append("*")
            continue

        # 6. Whiskers (Six straight horizontal/diagonal lines)
        # Left side whiskers
        if nx < -2:
            if abs(ny - 1.5 - 0.2*nx) < 0.2:  # Top-left whisker
                row_chars.append("*")
                continue
            if abs(ny + 1) < 0.2:             # Middle-left whisker
                row_chars.append("*")
                continue
            if abs(ny + 3.5 + 0.2*nx) < 0.2:  # Bottom-left whisker
                row_chars.append("*")
                continue
        # Right side whiskers
        if nx > 2:
            if abs(ny - 1.5 + 0.2*nx) < 0.2:  # Top-right whisker
                row_chars.append("*")
                continue
            if abs(ny + 1) < 0.2:             # Middle-right whisker
                row_chars.append("*")
                continue
            if abs(ny + 3.5 - 0.2*nx) < 0.2:  # Bottom-right whisker
                row_chars.append("*")
                continue

        # 7. Collar (Straight line at the bottom)
        if ny == -7 and -7.5 <= nx <= 7.5:
            row_chars.append("*")
            continue

        # 8. Bell (Small round circle dangling from the collar)
        bell_val = nx**2 + (ny + 8.5)**2
        if 0.6 <= bell_val <= 1.5:
            row_chars.append("*")
            continue

        # If no elements match, print an empty space
        row_chars.append(" ")
    print("".join(row_chars))
```

```python
import numpy as np

# Higher resolution grid for better detail
height = 40
width = 80

for y in range(height):
    row_chars = []
    for x in range(width):
        # Normalize coordinates to center (0, 0)
        # Using 1.85 to compensate for rectangular font aspect ratio in terminals
        nx = (x - 40) / 1.85
        ny = (18 - y)

        # 1. Main Head Outline (Radius approx 15)
        head_val = nx**2 + ny**2
        if 210 <= head_val <= 245:
            row_chars.append("*")
            continue

        # 2. Eyes (Left and Right Ellipses)
        left_eye = ((nx + 3.2)**2)/2.2 + ((ny - 7.5)**2)/3.5
        right_eye = ((nx - 3.2)**2)/2.2 + ((ny - 7.5)**2)/3.5

        if (0.8 <= left_eye <= 1.25) or (0.8 <= right_eye <= 1.25):
            row_chars.append("*")
            continue

        # Pupils (Inner dots inside the eyes)
        if (left_eye < 0.3) or (right_eye < 0.3):
            row_chars.append("*")
            continue

        # 3. Nose (Circle centered just under the eyes)
        nose_val = nx**2 + (ny - 3.8)**2
        if 1.0 <= nose_val <= 3.0:
            row_chars.append("*")
            continue

        # 4. Nose bridge / Vertical line from nose to mouth
        if abs(nx) < 0.45 and -7.0 <= ny <= 2.2:
            row_chars.append("*")
            continue

        # 5. Smiling Mouth (Symmetrical wide arc)
        mouth_val = nx**2 + (ny + 1.5)**2
        if 65.0 <= mouth_val <= 75.0 and ny < -1.5:
            row_chars.append("*")
            continue

        # 6. Whiskers (Six lines extending outward)
        # Left Whiskers
        if nx < -3.5:
            if abs(ny - 3.0 - 0.25*nx) < 0.25:  # Top-Left
                row_chars.append("*")
                continue
            if abs(ny + 1.0) < 0.25:             # Middle-Left
                row_chars.append("*")
                continue
            if abs(ny + 5.0 + 0.25*nx) < 0.25:  # Bottom-Left
                row_chars.append("*")
                continue
        # Right Whiskers
        if nx > 3.5:
            if abs(ny - 3.0 + 0.25*nx) < 0.25:  # Top-Right
                row_chars.append("*")
                continue
            if abs(ny + 1.0) < 0.25:             # Middle-Right
                row_chars.append("*")
                continue
            if abs(ny + 5.0 - 0.25*nx) < 0.25:  # Bottom-Right
                row_chars.append("*")
                continue

        # 7. Collar
        if ny == -11 and -11.5 <= nx <= 11.5:
            row_chars.append("*")
            continue

        # 8. Bell
        bell_val = nx**2 + (ny + 13.0)**2
        if 1.5 <= bell_val <= 4.0:
            row_chars.append("*")
            continue

        row_chars.append(" ")
    print("".join(row_chars))
```

```python
"""
Original sci-fi helmet drawn with Python turtle.
Spyder mein run karne ke liye: file open karo aur F5 dabao.
(Agar graphics window dikhe nahi, to Spyder: Preferences > IPython console > Graphics > Backend = "Automatic" ya "Inline" nahi; ya console ko "Run in external system terminal" pe chalao.)
"""

import turtle

# ---------- Setup ----------
screen = turtle.Screen()
screen.setup(700, 700)
screen.bgcolor("#0b1220")
screen.title("Sci-Fi Helmet")
screen.tracer(0)  # fast drawing

pen = turtle.Turtle()
pen.hideturtle()
pen.speed(0)


# ---------- Helpers ----------
def mirror(points):
    """Left-right mirror of a point list."""
    return [(-x, y) for x, y in points]


def poly(points, fill, outline="#e8f1f8", width=3):
    """Draw a filled polygon from a list of (x, y) points."""
    pen.penup()
    pen.goto(points[0])
    pen.color(outline, fill)
    pen.pensize(width)
    pen.pendown()
    pen.begin_fill()
    for p in points[1:]:
        pen.goto(p)
    pen.goto(points[0])
    pen.end_fill()
    pen.penup()


def line(p1, p2, color="#00e5ff", width=3):
    pen.penup()
    pen.goto(p1)
    pen.color(color)
    pen.pensize(width)
    pen.pendown()
    pen.goto(p2)
    pen.penup()


def dot(x, y, size, color):
    pen.penup()
    pen.goto(x, y)
    pen.dot(size, color)


# ---------- Colors ----------
SHELL = "#9aa7b4"
SHELL_DARK = "#6e7b88"
CYAN = "#00bcd4"
GLOW = "#00f0ff"
VISOR = "#06121f"

# ---------- Outer shell ----------
shell = [(0, 250), (-120, 232), (-190, 150), (-212, 20), (-192, -100),
         (-140, -190), (-60, -232), (60, -232), (140, -190), (192, -100),
         (212, 20), (190, 150), (120, 232)]
poly(shell, SHELL)

# ---------- Forehead plates ----------
poly([(-110, 200), (-20, 215), (-20, 130), (-130, 120)], SHELL_DARK, width=2)
poly(mirror([(-110, 200), (-20, 215), (-20, 130), (-130, 120)]), SHELL_DARK, width=2)

# ---------- Center crest ----------
crest = [(0, 252), (-26, 215), (-20, 118), (20, 118), (26, 215)]
poly(crest, CYAN, outline="#e8f1f8")
line((0, 240), (0, 125), color="#e8f1f8", width=2)

# ---------- Cheek plates ----------
cheek = [(-192, -10), (-142, -38), (-102, -118), (-142, -172), (-186, -100)]
poly(cheek, SHELL_DARK, width=2)
poly(mirror(cheek), SHELL_DARK, width=2)

# ---------- Visor ----------
visor = [(-152, 62), (-112, 102), (-40, 116), (0, 106), (40, 116), (112, 102),
         (152, 62), (142, -10), (92, -42), (30, -30), (0, -46), (-30, -30),
         (-92, -42), (-142, -10)]
poly(visor, VISOR, outline=CYAN, width=4)

# ---------- Glowing eyes ----------
eye = [(-122, 52), (-62, 72), (-40, 50), (-112, 28)]
poly(eye, GLOW, outline="#ffffff", width=2)
poly(mirror(eye), GLOW, outline="#ffffff", width=2)

# Visor shine lines
line((-95, 92), (-50, 104), color="#3b6a8a", width=2)
line((95, 92), (50, 104), color="#3b6a8a", width=2)

# ---------- Chin / mouth guard ----------
chin = [(-72, -108), (72, -108), (56, -192), (-56, -192)]
poly(chin, SHELL_DARK)

# Vent lines
for i in range(5):
    y = -125 - i * 13
    half = 58 - i * 3.5
    line((-half, y), (half, y), color=CYAN, width=3)

# ---------- Ear modules ----------
for side in (-1, 1):
    dot(side * 205, 15, 52, SHELL_DARK)
    dot(side * 205, 15, 34, CYAN)
    dot(side * 205, 15, 14, GLOW)

# ---------- Cyan trim lines on shell ----------
line((-150, 150), (-60, 235), color=CYAN, width=3)
line((150, 150), (60, 235), color=CYAN, width=3)
line((-60, -232), (-140, -190), color=CYAN, width=3)
line((60, -232), (140, -190), color=CYAN, width=3)

# ---------- Show ----------
screen.update()

try:
    screen.exitonclick()  # window pe click karo to band hogi
except turtle.Terminator:
    pass
```
