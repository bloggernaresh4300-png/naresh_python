```python
#a company makes a average profit of 100 million dollars with a standard deviation of 50 million dollars what is the
#probabliblity that the company will incur a loss in the next quarter ?
```

```python
import scipy.stats as stats

# Define the distribution parameters
mean_profit = 100  # in millions
std_dev = 50       # in millions

# A loss occurs when profit is less than 0
probability_of_loss = stats.norm.cdf(0, loc=mean_profit, scale=std_dev)

print(f"Probability of incurring a loss: {probability_of_loss:.4f} (or {probability_of_loss * 100:.2f}%)")
```
